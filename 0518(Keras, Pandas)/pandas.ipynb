```python
# 이 프로젝트의 목적 : '당뇨병 발병'을 예측하는 것
# 그렇다면, 모든 정보가 '당뇨병 발병'과 어떤 관계가 있는지
import pandas as pd
df = pd.read_csv('pima-indians-diabetes.csv', 
                  names = ["pregnant", "plasma", "pressure", "thickness", "insulin", "BMI", "pedigree", "age", "class"]) 

# group by
# as_index : pregnant 왼쪽 인덱스
# mean() : 평균 구하기, sort_values : 정렬
print( df[['pregnant', 'class']].groupby(['pregnant'], as_index=False).mean().sort_values(by='pregnant', ascending=True) )

# 통계적으로 봤을 때, 임신 횟수와 당뇨병 발병 횟수 간 상관관계가 있어 보인다.
# 히지만 이 데이터의 경우, 임신 안된 케이스에 남자도 포함되어 있다.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 12))

sns.heatmap(df.corr(), linewidths=0.1, vmax=0.5, cmap=plt.cm.gist_heat, linecolor='white', annot=True)

# df.corr() : Data
# linewidths : 각 셀을 구분할 선의 굵기
# vmax, vmin : 컬러 맵을 고정하는 값. 그렇지 않으면 데이터 및 다른 키워드 인수에서 유추.
# cmap : Color Map. 데이터 값에서 색 공간으로 매핑, 설정하지 않으면 기본값은 중심 설정 여부에 따라 다름.
# linecolor : 구분선 색상
# annot : Data 값을 각 셀에 표시

# class 1 기준, pregnant : 0.22, plasma : 0.47, pressure : 0.065, thickness : 0.075, insulin : 0.13, BMI : 0.29 ... age : 0.24
# 0.1 이상이면 상관관계가 있다고 판단한다.
# 상관관계표가 학습의 기준이 되지는 않는다.
```
