```python
import matplotlib as mpl
import matplotlib.pyplot as plt

import sys
sys.path.append("mat_kor.py")

import mat_kor
import numpy as np

## subplot()
x1 = np.linspace(0.0, 5.0) # np.linspace(시작 값, 끝 값, 갯수(default : 50))
x2 = np.linspace(0.0, 2.0)
y1 = np.cos(2 * np.pi * x1) * np.exp(-1)
y2 = np.cos(2 * np.pi * x2)

ax1 = plt.subplot(2, 1, 1) # 2행 1열 행렬의 첫번째 (위)
plt.plot(x1, y1, 'yo-')
plt.title("subplot1")
plt.ylabel("Damped oscillation")
print(ax1)

ax2 = plt.subplot(2, 1, 2) # 2행 1열 행렬의 두번째 (아래)
plt.plot(x2, y2, 'r.-')
plt.xlabel("time(s)")
plt.ylabel("undamped")
print(ax2)

plt.tight_layout() # 여백을 줄여줌
plt.show()
```

```python
## 여러가지 플롯을 하나의 Axes 객체에 표시할 때 y값의 크기가 달라서 표시하기 힘든 경우가 있다. 
## 이 때는 다음처럼 twinx 명령으로 복수의 y 축을 가진 플롯을 만들수도 있다. 
## twinx 명령은 x 축을 공유하는 새로운 Axes 객체를 만든다.
import matplotlib as mlb
import matplotlib.pyplot as plt

import sys
sys.path.append("mat_kor.py")

import mat_kor

fig, ax0 = plt.subplots()
ax1 = ax0.twinx() # x축을 공유하는 Axes

ax0.set_title("2개의 y축을 한 figure에서 사용한다.")
ax0.plot([10, 5, 2, 9, 7], 'r-', label="y0") #범례 y0
ax0.legend(loc=0)
ax0.set_ylabel("y0") #y축 이름
ax0.grid(True) # ax0.set_xlabel이 기반이기 때문에, 그리드가 x, y로 그려진다

ax1.plot([100, 200, 220, 180, 110], 'g:', label="y1") #범례 y1
ax1.legend(loc=0)
ax1.set_ylabel("y1") #y축 이름
ax1.grid(False)

ax0.set_xlabel("공유 x축") # twinx()

plt.show()
```
