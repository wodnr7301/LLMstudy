```python
import tensorflow as tf

#상수 정의
a = tf.constant(5)
b = tf.constant(2)
c = tf.constant(3)

#연산
d = tf.multiply(a, b)
e = tf.add(c, b)
f = tf.subtract(d, e)

#값 출력
print("a=", a.numpy())

#실제 숫자 값 출력
print("b=", b.numpy())
print("c=", c.numpy())
print("d=", d.numpy())
print("e=", e.numpy())
print("f=", f.numpy())
```

```python
import tensorflow as tf

#입력과 가중치 정의
I = tf.Variable([[1,2,3,4], [5,6,7,8]], dtype=tf.float32)
W = tf.Variable([[1,1], [2,2], [3,3], [4,4]], dtype=tf.float32)

#행렬곱 계산
m_cal = tf.matmul(I, W)

#결과 출력
print(m_cal.numpy())
```

```python
# 재할당

val = tf.Variable([1, 3, 5, 7, 9], dtype=tf.float32)
print(f"val before : {val.numpy()}")

# val.assign([3, 6, 9]) : x -> Variable은 처음 생성될 때 고정된 크기, 고정된 데이터 타입.
# val.assign([3, 6, 9, 12, 15], dtype=tf.int) : x
val.assign([3, 6, 9, 12, 15])
print(f"val after : {val.numpy()}")
```

```python
# 브로드 캐스팅
# x : (2x3), W : (3x1)
#앞 행렬의 열 갯수 * 뒤 행렬의 행 갯수가 일치하므로 계산 가능
# +b에서 브로드 캐스팅
# [6], [12] (2x1)
# b : [4] (1X1) 행렬곱은 가능하지만, +b이기 때문에 행과 열이 동일해야 한다.
# 브로드 캐스팅은 [4], [4] 로 만들어서 덧셈이 가능하게 해준다.

import tensorflow as tf

#입력데이터
x = tf.constant([[1, 1, 1], [2, 2, 2]], dtype=tf.float32)

#가중치, bias
W = tf.Variable([[2], [2], [2]], dtype=tf.float32)
b = tf.Variable([4], dtype=tf.float32)

y = tf.matmul(x, W)
print(f" 브로드캐스팅 전 b : {y.numpy()}")

y = y + b
print(f" 브로드캐스팅 후 b : {y.numpy()}")
```
