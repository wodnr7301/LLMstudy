```python
#딥러닝을 구동하는데 필요한 케라스 함수를 불러옵니다.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

#필요한 라이브러리를 불러옵니다.
import numpy as np
import tensorflow as tf

#실행할 떄마다 같은 결과를 출력하기 위해 설정하는 부분입니다.
np.random.seed(3)
tf.random.set_seed(3)

#준비된 수술 환자 데이터를 불러들입니다.
Data_set = np.loadtxt("ThoraricSurgery.csv", delimiter=",")

#환자의 기록과 수술결과를 X와 Y로 구분하여 저장합니다.
X = Data_set[:, 0:17]
Y = Data_set[:, 17]

#딥러닝 구조를 결정합니다.
model = Sequential()
model.add(Input(shape=(17,))) # 2.x 후반에 변경된 부분 : input 레이어 사용
model.add(Dense(30, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

#딥러닝을 실행합니다.
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X, Y, epochs=100, batch_size=10)

#결과를 출력합니다.
print("\n Accuracy : %.4f" % (model.evaluate(X, Y)[1]))
```

```python
import numpy as np

# 데이터 정의(입력 데이터, 출력 데이터)
data_x = np.array([10, 20, 30, 40])
data_y = np.array([21, 49, 68, 80])

# 평균값 계산
mean_x = np.mean(data_x)
mean_y = np.mean(data_y)

# 기울기 계산
numerator = np.sum((data_x-mean_x)*(data_y-mean_y))
denominator = np.sum((data_x-mean_x)**2)
slope = numerator/denominator

# 절편 계산
imtercept = mean_y - slope * mean_x

# 결과 출력
print("기울기", slope)
print("절편", imtercept)
```
