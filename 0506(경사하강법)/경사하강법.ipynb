```python
#답
import numpy as np

#실제값과 예측값
y_true = [21, 49, 68, 80]
y_pred = [25.1, 44.7, 64.3, 83.9]

#RMSE계산 함수
def rmse(y_true, y_pred):
    mse = np.mean((np.array(y_true) - np.array(y_pred))**2)
    return np.sqrt(mse)

#결과출력
print("RMSE : ", rmse(y_true, y_pred))
```

```python
#연습
import numpy as np

#데이터
x_data = np.array([1,2,3,4,5,6,7,8])
y_data = np.array([0.2, 0.3, 0.5, 0.6, 0.9, 0.95, 1.1, 1.5])

#평균
x_mean = np.mean(x_data)
y_mean = np.mean(y_data)
print(f"x_mean : {x_mean}")
print(f"y_mean : {y_mean}")

#기울기
numerator = np.sum((x_data - x_mean)*(y_data - y_mean))
denominator = np.sum((x_data - x_mean)**2)
slope = numerator/denominator
print(f"slope : {slope}")

#편차
interceptor = y_mean - (slope * x_mean)
print(f"interceptor : {interceptor}")

#예측 식
y_pred = np.array((slope * x_data + interceptor))
print(f"y_pred : {y_pred}")

#오차
mse = np.mean((y_data - y_pred)**2)
rmse = np.sqrt(mse)
print(f"mse : {mse.round(3)}")
print(f"rmse : {rmse.round(4)}")
```

```python
# 답
import numpy as np

# 사육 기간(주차)와 실제 크기(cm)
weeks = np.array([1, 2, 3, 4, 5, 6, 7, 8])
actual_sizes = np.array([0.2, 0.3, 0.5, 0.6, 0.9, 0.95, 1.1, 1.5])

# 예측 식: y = 0.1756x - 0.03392
predicted_sizes = 0.1756 * weeks - 0.03392

# 오차 및 RMSE 계산
errors = actual_sizes - predicted_sizes
squared_errors = errors ** 2
mse = np.mean(squared_errors)
rmse = np.sqrt(mse)

# 출력
print("실제 크기:", actual_sizes)
print("예측 크기:", predicted_sizes.round(3))
print("오차:", errors.round(3))
print("제곱 오차:", squared_errors.round(4))
print(f"\n최종 RMSE: {rmse:.10f}")
```

```python
import numpy as np
# import pandas as pd
import matplotlib.pyplot as plt

#공부 시간 x와 성적 y의 리스트를 만들기
data = [[2, 81], [4, 93], [6, 91], [8, 97]]
x = [i[0] for i in data]
y = [i[1] for i in data]

#x, y data확인
print(x, y)

#그래프로 나타내기
plt.figure(figsize=(8,5))
plt.scatter(x, y)
plt.show()

#리스트로 되어있는 x와 y값을 넘파이 배열로 바꾸기
x_data = np.array(x)
y_data = np.array(y)

#기울기 a와 절편 b 초기화
a=0
b=0

#학습률
lr=0.03

#몇번 반복될지 설정
#0부터 세므로 원하는 반복횟수에 +1
epochs = 2001

#경사 하강법 시작
for i in range(epochs) :
    a_diff = 2/len(x_data) * sum((a * x_data + b - y_data)*x_data)
    b_diff = 2/len(x_data) * sum(a * x_data + b - y_data)

    a = a-lr*a_diff
    b = b-lr*b_diff
    if i&100==0 :
        print("epoch=%.f, 기울기=%.04f, 절편=%.04f" % (i, a, b))

#앞서 구한 기울기와 절편을 이용해 그래프를 다시 그리기
y_pred = a * x_data + b
plt.scatter(x,y)
#plt.plot([min(x_data), max(x_data)], [min(y_pred), max(y_pred)])
plt.plot(x_data,y_pred)
plt.show()
```
